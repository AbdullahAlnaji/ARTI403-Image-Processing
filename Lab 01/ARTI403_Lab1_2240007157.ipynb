## Task #1: Exercise with Spyder

```python
# Quick warm-up exercise using Python/NumPy basics
import numpy as np

# simple list and loop practice
numbers = list(range(1, 11))
squares = [n**2 for n in numbers]
print("Numbers:", numbers)
print("Squares:", squares)

# simple NumPy array practice
arr = np.array(numbers)
print("Array:", arr)
print("Mean:", arr.mean(), "| Std:", arr.std())
```

## Task #2: Loading and Visualizing Images

### a. Loading/reading an image using OpenCV

```python
import cv2
img = cv2.imread('cameraman.jpg')  # Load the image data into a NumPy array

import matplotlib.pyplot as plt
plt.imshow(img)
plt.title('Cameraman - loaded with OpenCV')
plt.show()
```

### b. Loading/reading image using PIL

```python
from PIL import Image
img2 = Image.open('lena_gray_256.webp') #load image into a PIL Image object

import matplotlib.pyplot as plt
import matplotlib.cm as cm
plt.imshow(img2, cmap = cm.Greys_r)
plt.show()
```

## Task #3: Image Storing

### a. Saving image to disk using OpenCV

```python
cv2.imwrite('new_image.jpg', img)
print("Saved new_image.jpg using OpenCV")
```

### b. Saving an image to disk using PIL

```python
img2.save('new_image2.jpg')
print("Saved new_image2.jpg using PIL")
```

## Task #4: Display Image as an Array

```python
print(img.shape)  # Note: img loaded using OpenCV is a NumPy array
print(img)

import numpy as np

# Convert the PIL image to a NumPy array
img_array = np.array(img2)

# Print the array
print(img_array.shape)
print(img_array)
```

## 4. Assessment
Try performing different operations on `img_array` using NumPy, for example:

```python
# Example assessment operations on the array
print("Min pixel value:", img_array.min())
print("Max pixel value:", img_array.max())
print("Mean pixel value:", img_array.mean())

# Slice out the top-left 50x50 corner of the image
corner = img_array[:50, :50]
print("Corner shape:", corner.shape)

# Flip the image array horizontally
flipped = np.fliplr(img_array)
plt.imshow(flipped, cmap=cm.Greys_r)
plt.title('Flipped image (NumPy)')
plt.show()
```
